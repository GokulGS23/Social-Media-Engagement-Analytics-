```python
df["age"]=df["age"].fillna(df["age"].median())

df["likes"]=df["likes"].fillna(df["likes"].median())

df["comments"]=df["comments"].fillna(df["comments"].median())

df["shares"]=df["shares"].fillna(df["shares"].median())
```

```python
df["gender"]=df["gender"].fillna(df["gender"].mode()[0])
df["sentiment"]=df["sentiment"].fillna(df["sentiment"].mode()[0])
```

# Data Visualization 

## Likes vs Impressions

```python
import matplotlib.pyplot as plt
plt.scatter(df["impression_count"],df["likes"])
plt.title("Likes vs Impressions")
plt.xlabel("Impressions")
plt.ylabel("Likes")
plt.show()
```

## daily engagement trend

```python
daily_engagement =(df.groupby(df["posted_at"].dt.date)["engagement_score"].mean())

plt.plot(daily_engagement.index.astype(str),daily_engagement.values)
plt.title("Daily Engagement Trend")
plt.xlabel("Date")
plt.ylabel("Average Engagement Score")

plt.show()
```

## posts by category

```python
category_counts =df["post_category"].value_counts()
plt.bar(category_counts.index,category_counts.values)
plt.title("Number of posts by category")
plt.xlabel("Post Category")
plt.ylabel("Number of Posts")
plt.show()
```

## gender distribution

```python
gender_counts = df["gender"].value_counts()
plt.pie(gender_counts.values,labels=gender_counts.index)
plt.title("Gender Distribution")
plt.show()
```

## age
 


```python
plt.hist(df["age"],bins=20)
plt.title("Age Distribution")
plt.xlabel("Age")
plt.ylabel("Frequency")
plt.show()
```

## engagement rate

```python
plt.boxplot(df["engagement_rate"])
plt.title("Engagement Rate Distribution")
plt.ylabel("Engagement Rate")
plt.show()
```
